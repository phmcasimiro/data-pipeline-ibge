# NumPy

## 📦 1. O que é NumPy?

NumPy = **Numerical Python**. Ela introduz o objeto `ndarray` (N-dimensional array), que é:
- Mais eficiente que listas (em memória e velocidade)
- Vetorizado (operações em todo o array de uma vez)
- Base para álgebra linear, estatística, etc.

Para usar, importe com o alias padrão:


````python
import numpy as np
````

## 🏗️ 2. Criando Arrays

### 2.1 `np.array()` – a partir de uma lista/tupla

````python
# 1D (vetor)
arr1 = np.array([1, 2, 3, 4, 5])
print(arr1)
````

````python
# 2D (matriz)
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
print(arr2)
````

````python
# 3D (tensor)
arr3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(arr3)
````

### 2.2 `np.zeros()` – array preenchido com zeros

````python
# zero 1D com 5 elementos
z = np.zeros(5)
print(z)
````

````python
# zero 2D (3 linhas, 4 colunas)
z2 = np.zeros((3, 4))
print(z2)
````

### 2.3 `np.ones()` – array preenchido com uns

````python
o = np.ones((2, 3))
print(o)
````

### 2.4 `np.arange()` – sequência numérica (similar ao `range()`)

````python
# De 0 a 9 (step padrão = 1)
a = np.arange(10)
print(a)
````

````python
# De 2 a 10 (exclusive), step 2
b = np.arange(2, 11, 2)
print(b)
````

````python
# Com floats
c = np.arange(0, 1, 0.2)
print(c)
````

### 2.5 Bônus: `np.linspace()` – pontos igualmente espaçados

````python
# 5 pontos entre 0 e 10 (inclusive)
lin = np.linspace(0, 10, 5)
print(lin)
````

## 📐 3. Shape (Formato)

O **shape** é uma tupla que indica o número de elementos em cada dimensão.

````python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
print(arr.shape)  # (2, 3) → 2 linhas, 3 colunas
````

````python
# Para 1D:
v = np.array([1, 2, 3])
print(v)
print(v.shape)    # (3,) → tupla com 1 elemento (atenção à vírgula!)
````

````python
# Para escalar (0D):
# cria um array numpy com o número inteiro 42
# Como é passado apenas um número (escalar), o array resultante possui 0 dimensões, ou seja, um ponto no espaço matemático
s = np.array(42)
print(s)
print(s.shape)    # () → tupla vazia
````

### 📌 Regras importantes:

- **len(shape)** = número de dimensões (ndim)

- O produto dos elementos do shape = total de elementos (size)

````python
arr1 = np.array([[1, 2, 3], [4, 5, 6]])

print(arr1.shape)  # (2,3)
print(arr1.ndim)   # 2
print(arr1.size)   # Quantidade de elementos do array/matriz: (6) ou (2*3) 
````

### 🔄 Redimensionar (reshape)

- Método RESHAPE é utilizado para mudar a forma, isto é, reorganizar os elementos em uma nova estrutura de array/matriz

````python
arr2 = np.arange(12)  # [0 1 2 ... 11] shape (12,)
print(arr2) # Exibe a organização do array
print(arr2.shape) # Exibe a quantidade de linhas e colunas (12 linhas, sem colunas/vazio)
print(arr2.ndim) # Exibe a quantidade de dimensões (1)
print(arr2.size) # Exibe a quantidade de elementos do array (12)
````

````python
# Transformar em 3x4
arr_3x4 = arr2.reshape(3, 4)
print(arr_3x4)# Exibe a nova organização do array
print(arr_3x4.shape)  # Exibe a quantidade de linhas e colunas (3 linhas, 4 colunas)
print(arr_3x4.ndim) # Exibe a quantidade de dimensões
print(arr_3x4.size) # Exibe a quantidade de elementos
````

````python
# Reshape/Transformar em 3 dimensões 2x2x3
# Reshape (bloco, linha, coluna)

arr_2x2x3 = arr2.reshape(2, 2, 3)
print(arr_2x2x3) # Exibe a nova organização do array
print(arr_2x2x3.shape)  # Exibe a quantidade de blocos (d1) x linhas (d2) x colunas (d3) (2, 2, 3)
print(arr_2x2x3.ndim) # Exibe a quantidade de dimensões
print(arr_2x2x3.size) # Exibe a quantidade de elementos
````

````python
# RESHAPE com cálculo automático de linhas e colunas
# arr2.reshape(2, -1)
# arr2.reshape(2 linhas, calcule colunas)
# Usar -1 para inferir automaticamente
arr_auto = arr2.reshape(2, -1)
print(arr_auto) # Exibe a nova organização do array
print(arr_auto.shape)  # Exibe a quantidade de linhas (2) x colunas calculadas automaticamente (6)
print(arr_auto.ndim) # Exibe a quantidade de dimensões
print(arr_auto.size) # Exibe a quantidade de elementos
````

## 🏷️ 4. dtype (Tipo de Dado)

O **dtype** define como os dados são armazenados na memória (inteiro, float, string, etc.).

- O Numpy infere automaticamente o dtype, mas pode ser forçado.


### 📌 Dtypes mais comuns:

| dtype | Descrição |
|-------|-----------|
| `int32`, `int64` | Inteiros com sinal |
| `uint8`, `uint16` | Inteiros sem sinal (0 a 255, etc) |
| `float32`, `float64` | Ponto flutuante |
| `bool` | Booleano (True/False) |
| `str` | String (menos eficiente) |
| `complex128` | Números complexos |

````python
arr_int = np.array([1, 2, 3]) # array de inteiros
print(arr_int)
print(arr_int.dtype)

arr_float = np.array([1.0, 2.5, 3.7]) # array de float64
print(arr_float)
print(arr_float.dtype)

arr_forced = np.array([1, 2, 3], dtype=np.float32) # forçando tipo float32 em um array de inteiros
print(arr_forced)
print(arr_forced.dtype)


arr_mixed = np.array([1, 2, 3.5]) # Array de inteiros que se torna float se misturar
print(arr_mixed)
print(arr_mixed.dtype)  # float64 (por causa do 3.5)
````

### Especificando dtype nas funções de criação:

````python
zeros_float = np.zeros(5, dtype=np.float32)
print(zeros_float)
print(zeros_float.dtype)
print()
ones_int = np.ones((2, 3), dtype=np.int8)
print(ones_int)
print(ones_int.dtype)
print()
arange_float = np.arange(0, 10, dtype=np.float64)
print(arange_float)
print(arange_float.dtype)
````


## 🧠 6. Dicas importantes

1. **Arrays vs Listas:**
   - Arrays são homogêneos (todos os elementos do mesmo tipo)
   - Se misturar tipos, NumPy faz *upcasting* (int → float → str)

2. **Cópia vs View:**
   - `reshape()` retorna uma *view* (não copia dados)
   - Use `.copy()` se precisar de uma cópia independente

3. **Performance:**
   - Operações com arrays são em C (muito mais rápidas que loops Python)
