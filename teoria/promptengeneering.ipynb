# PROMPT ENGINEERING

## FUNDAMENTOS

### 1. O que é um Prompt?
Um prompt é uma instrução de entrada. Os modelos de IA (LLMs) são como um músico extremamente talentoso, mas que precisa de uma partitura para tocar a música que o ouvinte quer ouvir. Se a instrução for "toque algo", o músico vai improvisar de qualquer jeito. Mas se na instrução for definida a partitura desejada, o músico vai tocar exatamente a música desejada.

---

### 2. Os 3 Pilares do Prompt Perfeito
Para construir um bom prompt, deve ser usada a regra dos 3 "C" + 1 "E":

- **CONTEXT (Contexto):** Onde estamos? Qual é o cenário?
- **COMMAND (Instrução):** O que você quer que eu faça exatamente? (Verbo no imperativo: "Liste", "Explique", "Traduza", "Código").
- **CHARACTERISTIC(Características):** Como deve ser a saída? (Tom, formato, tamanho, público-alvo).
- **EXAMPLE (Exemplo):** *Opcional, mas poderoso.* Mostre um exemplo do que você quer.

---

### 3. Técnicas de Prompt Engineering

#### Técnica 1: Zero-shot vs. Few-shot
- **Zero-shot:** Pedir algo sem dar exemplos. 
- **Few-shot:** Dar 1, 2 ou 3 exemplos antes de pedir a tarefa real. Isso "ensina" o padrão que a IA deve seguir.

#### Técnica 2: Chain-of-Thought (CoT) - Cadeia de Pensamento
É pedir para a IA "pensar em voz alta", mostrando o raciocínio passo a passo. Isso reduz alucinações e melhora a matemática e a lógica.

#### Técnica 3: Persona (Atribuição de Papel)
Mandamos a IA assumir um papel específico (médico, advogado, programador sênior, professor chato de IA). Isso muda o vocabulário e a abordagem.

#### Técnica 4: Restrição de Formato (Formatting)
Forçar a saída a ser em JSON, Markdown, tabela, ou com um número exato de parágrafos.

#### **Cenário 3: O Poder do Few-Shot + Chain-of-Thought (Cadeia de Pensamento)**

*prompt (Excelente):* 
> "Preciso que você classifique o sentimento das frases dos clientes como 'Positivo', 'Negativo' ou 'Neutro'. 
> 
> Exemplo 1: 'A entrega foi extremamente rápida!' -> Positivo
> Exemplo 2: 'O produto veio quebrado e o suporte não respondeu.' -> Negativo
> 
> Agora, analise esta frase: 'O aplicativo até que funciona, mas a bateria do celular descarregou muito rápido enquanto eu usava.' 
> 
> Antes de me dar a resposta final, mostre seu raciocínio passo a passo sobre por que escolheu essa classificação."

*resposta (Excelente):*
"**Raciocínio (Chain-of-Thought):**
1. A frase menciona que o aplicativo 'até que funciona' – isso indica que a função principal é aceitável, o que é um ponto positivo.
2. No entanto, a conjunção 'mas' introduz uma reclamação forte sobre a bateria descarregar muito rápido.
3. O problema da bateria é um efeito colateral negativo que prejudica a experiência do usuário.
4. Peso dos fatores: O ponto negativo (bateria) é mais impactante que o ponto positivo (funciona) porque causa um dano real ao usuário.

**Classificação Final:** Negativo."

## PROMPT EM CAMADAS - Método GOAL com Claude

### G - Goal / Objetivo
- O que deve ser alcançado de modo mais claro ou específico possível
  - Por exemplo, uma landing page para e-commerce de roupas femininas

### O - Output / Saída
- Como a resposta deve ser entregue, ou seja, o formato de saída
  - Por exemplo, uma página html que utilize html, css e javascript, responsivo e moderno

### A - Audience
- Qual o público alvo da resposta, a quem deve ser direcionada
  - Por exemplo, mulheres entre 20 e 30 anos interessadas em moda

### L - Layers
- detalhes adicionais que enriquecem o prompt, contexto, estilos, restrições, regras
  - cabeçalho com logo e menu de navegação;
  - banner principal com chamada para ação;
  - seção de produtos com imagens, preços e botão de compra;
  - seção de depoimentos de clientes;
  - Rodapé com contato e redes sociais;
  - ao clicar em comprar, alerta de "produto adicionado ao carrinho!"
