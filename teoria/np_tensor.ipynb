## 🧊 O que é um Tensor?

De forma simples: **um tensor é uma generalização de matrizes para dimensões mais altas**.

| Nome | Dimensões | Exemplo no NumPy |
|------|-----------|------------------|
| **Escalar** | 0D (sem dimensões) | `np.array(42)` → shape `()` |
| **Vetor** | 1D (uma dimensão) | `np.array([1,2,3])` → shape `(3,)` |
| **Matriz** | 2D (duas dimensões) | `np.array([[1,2],[3,4]])` → shape `(2,2)` |
| **Tensor** | 3D ou mais | `np.array([[[1,2],[3,4]],[[5,6],[7,8]]])` → shape `(2,2,2)` |

---

## 📊 Analogia Visual

Pense em termos de "pilhas" de dados:

```
Escalar (0D):     [42]
                    ↑
                  um número

Vetor (1D):       [1, 2, 3, 4]
                    ↑
                  uma linha

Matriz (2D):      [[1, 2, 3],
                   [4, 5, 6]]
                    ↑
                  uma tabela (linhas × colunas)

Tensor 3D:        [[[1, 2],
                    [3, 4]],
                                        ↑
                   [[5, 6],          várias matrizes
                    [7, 8]]]         empilhadas
```

## 🎯 Exemplos Práticos no NumPy

### Criando Tensores 3D

```python
import numpy as np

# Tensor 3D: 2 matrizes, cada uma com 3 linhas e 4 colunas
tensor_3d = np.array([
    [[1, 2, 3, 4],   # matriz 1 (linha 0)
     [5, 6, 7, 8],   # matriz 1 (linha 1)
     [9, 10, 11, 12]], # matriz 1 (linha 2)
    
    [[13, 14, 15, 16], # matriz 2 (linha 0)
     [17, 18, 19, 20], # matriz 2 (linha 1)
     [21, 22, 23, 24]] # matriz 2 (linha 2)
])

print(tensor_3d.shape)  # (2, 3, 4) → 2 matrizes, 3 linhas, 4 colunas
print(tensor_3d.ndim)   # 3 (três dimensões)
```

### Tensor 4D (comum em Deep Learning)

```python
# Imagem colorida: [batch, altura, largura, canais]
# Exemplo: 10 imagens RGB de 64x64 pixels
tensor_4d = np.random.rand(10, 64, 64, 3)
print(tensor_4d.shape)  # (10, 64, 64, 3)
#          ↑    ↑    ↑    ↑
#        batch  altura largura canais (R,G,B)
```

## 🧮 Notação Matemática

Em matemática, usamos índices para representar elementos:

| Tipo | Notação | Exemplo |
|------|---------|---------|
| Escalar | `x` | `x = 5` |
| Vetor | `xᵢ` | `x = [1, 2, 3]` → `x₁=1, x₂=2, x₃=3` |
| Matriz | `Xᵢⱼ` | `X = [[1,2],[3,4]]` → `X₁₁=1, X₁₂=2` |
| Tensor 3D | `Tᵢⱼₖ` | `T[0,1,2]` acessa elemento específico |

## 💻 Acessando Elementos em Tensores

```python
tensor = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

# Acessando elementos
print(tensor[0, 0, 0])  # 1 (primeira matriz, primeira linha, primeira coluna)
print(tensor[1, 0, 1])  # 6 (segunda matriz, primeira linha, segunda coluna)

# Slicing em múltiplas dimensões
print(tensor[:, 0, :])  # Pega todas as matrizes, linha 0, todas as colunas
# [[1 2]
#  [5 6]]
```

## 🔍 Por que "Tensor" e não apenas "Array Multidimensional"?

O termo "tensor" vem da matemática e física (Einstein, relatividade geral) e implica **propriedades específicas de transformação**:

1. **Transformação sob mudança de base**: Como os valores se comportam quando giramos o sistema de coordenadas
2. **Invariância**: Algumas propriedades permanecem constantes independente da orientação

Na prática do dia a dia em Python, **usamos "tensor" e "array multidimensional" como sinônimos**, especialmente no contexto de aprendizado de máquina.

## 📐 Operações com Tensores

```python
# Criando tensores
A = np.random.rand(2, 3, 4)
B = np.random.rand(2, 3, 4)

# Operações elemento a elemento (broadcasting)
C = A + B          # soma
D = A * B          # multiplicação elemento a elemento
E = np.dot(A, B.T) # produto matricial (requer compatibilidade)

# Redução
soma_total = np.sum(A)           # soma todos elementos
media_por_matriz = np.mean(A, axis=0)  # média ao longo do primeiro eixo
max_por_linha = np.max(A, axis=2)      # máximo ao longo do último eixo
```

## 🧪 Exercício Prático

```python
import numpy as np

# 1. Crie um tensor 3D com shape (3, 4, 5) preenchido com números aleatórios
tensor = np.random.rand(3, 4, 5)

# 2. Descubra:
print(f"Shape: {tensor.shape}")      # (3, 4, 5)
print(f"Dimensões: {tensor.ndim}")   # 3
print(f"Total elementos: {tensor.size}")  # 60

# 3. Acesse o elemento da segunda matriz, terceira linha, quarta coluna
elemento = tensor[1, 2, 3]
print(f"Elemento [1,2,3]: {elemento}")

# 4. Some todos os elementos ao longo do primeiro eixo
soma_axis0 = np.sum(tensor, axis=0)
print(f"Soma ao longo do eixo 0: {soma_axis0.shape}")  # (4, 5)

# 5. Remodele para 2D
tensor_2d = tensor.reshape(6, 10)
print(f"Remodelado: {tensor_2d.shape}")  # (6, 10)
```

## 🎯 Resumo Final

| Conceito | Definição | No NumPy |
|----------|-----------|----------|
| **Tensor** | Array multidimensional generalizado | `ndarray` com ndim ≥ 3 |
| **Ordem** | Número de dimensões | `array.ndim` |
| **Shape** | Tamanho em cada dimensão | `array.shape` |
| **Uso comum** | Deep Learning, física, engenharia | Dados estruturados |

**Regra de ouro**: Se você consegue descrever seus dados com "uma tabela de tabelas de tabelas...", você está lidando com um tensor! 🚀
