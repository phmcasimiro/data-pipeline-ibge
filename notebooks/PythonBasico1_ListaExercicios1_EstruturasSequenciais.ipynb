#### 5. Faça um Programa que converta metros para centímetros. Você pode pedir o comprimento em metros para o usuário (input).

#### Versão 1: Com todas as condições (mais completa)

```python
metros = float(input("Medida em metros: "))

centimetros = metros * 100

if metros < 0:
    print('Erro: Não pode ser calculado (valor negativo)')
elif metros == 0:
    print('0 metros é equivalente a 0 centímetros.')
elif metros == 1:
    print(f'{metros} metro é equivalente a {centimetros} centímetros.')
else:
    print(f'{metros} metros são equivalentes a {centimetros} centímetros.')
```

#### Versão 2: Com validação de entrada

```python
try:
    metros = float(input("Medida em metros:"))

    if metros < 0:
        print('Erro: Valor negativo não pode ser calculado')

    else:
        centimetros = metros * 100

        if metros ==1:
            print(f'{metros:.0f} metro é equivalente a {centimetros:.0f} centímetros.')
        else:
            print(f'{metros:.0f} metros são equivalentes a {centimetros:.0f} centímetros.')

except ValueError:
    print('Erro: Digite um número válido!')
```

#### Versão 3: Usando função

```python
def converter_metros_para_centimetros(metros):
    if metros < 0:
        return "Erro: Valor negativo não pode ser calculado!"
    
    centimetros = metros * 100

    if metros == 1:
        return f'{metros:.0f} metro é equivalente a {centimetros:.0f} centímetros'

    else:
        return f'{metros:.0f} metros é equivalente a {centimetros:.0f} centímetros'

try:
    metros = float(input("Medida em metros: "))
    resultado = converter_metros_para_centimetros(metros)
    print(resultado)
except ValueError:
    print('Digite um número válido!')
```

#### 6. Faça um Programa que calcule a área de uma sala de um apartamento. Para isso, o seu programa precisa pedir a largura da sala, o comprimento da sala e imprimir a área em m² da sala.

```python
def calcular_area(largura, comprimento):
    
    if largura < 0:
        raise ValueError ("Erro: Largura não pode ser valor negativo")
    if comprimento < 0:
        raise ValueError ("Erro: comprimento não pode ser valor negativo")
    if largura == 0 or comprimento == 0:
        raise ValueError ("Erro: Dimensões não podem ser zero")

    area = largura * comprimento
    return area

try:
    largura = float(input('Informe a Largura:'))
    comprimento = float(input('Informe a comprimento:'))
    area = calcular_area(largura,comprimento)
    print(f'A área é {area:.0f} m²')

except ValueError as erro:
    print(f'Erro: {erro}')
```

#### 7. Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês. Calcule e mostre o total do seu salário no referido mês.

```python
def calcular_salario(valor,horas):
    if valor <= 0:
        raise ValueError("Erro: Valor de hora trabalhada não pode ser igual ou inferior a zero!")
    if horas < 0:
        raise ValueError("Erro: Quantidade de horas trabalhadas não pode ser igual ou inferior a zero!")

    remuneracao = valor * horas
    return remuneracao

try:
    valor = float(input('Informe o valor da hora de trabalho: '))
    horas = float(input('Informe a quantidade de horas trabalhadas: '))
    remuneracao = calcular_salario(valor,horas)
    print(f'Sua remuneração esse mês é: R$ {remuneracao:.0f}')
except ValueError as erro:
    print(f'Erro: {erro}')
```

#### 8. Vamos criar um conversor de temperatura. Faça um Programa que peça a temperatura em graus Fahrenheit, transforme e mostre a temperatura em graus Celsius.
$C = \frac{5}{9}(F-32)$

```python
def conversao_temp(temp_f):
    
    celsius = 5/9*(temp_f - 32)
    return celsius

try:
    temp_f = float(input('Informe a temperatura em Fahrenheit:'))
    temp_c = conversao_temp(temp_f)
    print(f'A temperatura em Graus Celsius é: {temp_c:.1f}')
except ValueError:
    print('Erro: Digite um valor válido!')
```

#### 10. Tendo como dados de entrada a altura (h) de uma pessoa, construa um algoritmo que calcule seu peso ideal, usando a seguinte fórmula:
$P = 72,7h - 58$

Lembrando que "algoritmo" nada mais é do que um programa, como todos os outros que você vem fazendo

```python
def peso_ideal(altura_cm):
    
    if altura_cm <= 0:
        raise ValueError('A altura não pode ser negativa ou zero')

    if altura_cm > 210:
        raise ValueError ('A altura não pode ser maior do que 210 cm')
    
    peso = 0.727 * altura_cm - 58
    return peso

try:
    altura_input = input('Informe a altura em centímetros:')
    # verifica a existência de ponto ou vírgula na string de entrada e sobe erro se houver
    if '.' in altura_input or ',' in altura_input:
        raise ValueError ('Informar altura em centímetros com número inteiro! Exemplo: 175')
    # Transforma o input da altura em formato string para formato inteiro 
    altura_cm = int(altura_input)
    peso = peso_ideal(altura_cm)
    print(f'O peso ideal para {altura_cm}cm é {peso:.1f}')

except ValueError as erro:
    print(f'Erro: {erro}')
```

#### 12. Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês.

#####  Calcule o salário bruto (horas * salario por hora)

##### Calcule o desconto do IR (11% do salário bruto)

##### Calcule o desconto do INSS (8% do salário bruto)

##### Calcule o salário líquido

```python
def calcular_salario(valor_h,horas_t_m):
    if valor_h <= 0:
        raise ValueError("Erro: Valor de hora trabalhada não pode ser igual ou inferior a zero!")
    if horas_t_m < 0:
        raise ValueError("Erro: Quantidade de horas trabalhadas não pode ser inferior a zero!")

    ir = 1 - 0.11
    inss = 1 - 0.08
    sindicato = 1 - 0.05
    remuneracao_bruta = (valor_h * horas_t_m)
    remuneracao_ir = remuneracao_bruta * ir
    remuneracao_ir_inss = remuneracao_ir * inss
    remuneracao_ir_inss_sind = remuneracao_ir_inss * sindicato
    remuneracao_liquida = remuneracao_ir_inss_sind
    return remuneracao_liquida

try:
    valor_h = float(input('Informe o valor da hora de trabalho: '))
    horas_t_m = float(input('Informe a quantidade de horas trabalhadas: '))
    salario = calcular_salario(valor_h,horas_t_m)
    print(f'Seu Salário esse mês é: R$ {salario:.0f}')
except ValueError as erro:
    print(f'Erro: {erro}')
```
