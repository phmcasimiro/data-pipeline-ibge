```python
faturamento = 10000
custo = 6000
lucro = faturamento - custo

# MÉTODO CONCATENAÇÃO
print('O faturamento da loja foi ' + str(faturamento) + ' reais')

# MÉTODO FORMAT
print('O faturamento da loja foi {}. O custo da loja foi {}. O lucro da loja foi {}'.format(faturamento, custo, lucro))

# MÉTODO F-STRING
print(f'O faturamento da loja foi {faturamento}. O custo da loja foi {custo}. O lucro da loja foi {lucro}')
```

## MUDANÇA DE TIPO DE VARIÁVEL

```python
# CÉLULA COM ERRO PROPOSITAL

# Este input faz a variável ser do tipo string
faturamento2 = input('Insira o faturamento')

# Este input faz a variável ser do tipo string
custo2 = input('Insira o custo')

# Esta conta não pode ser feita porque faturamento2 e custo2 são variáveis do tipo string;
# Para fazer essa conta é necessário taxar as variáveis faturamento2 e custo2 com tipo numérico (int ou float)
lucro2 = faturamento2 - custo2

print(lucro2)

# VERIFICAR O ERRO: TypeError: unsupported operand type(s) for -: 'str' and 'str'
```

```python
# SOLUÇÃO PARA O PROBLEMA ACIMA

# Forçar uma variável a ser do tipo float
faturamento2 = float(input('Insira o faturamento'))

# Forçar uma variável a ser do tipo float
custo2 = float(input('Insira o custo'))

lucro2 = faturamento2 - custo2

print(lucro2)
```

## CONDIÇÕES - if / else

```python
meta = 50000
qtd_venda = 45300

if qtd_venda >= meta:
    print('Meta batida!')
else:
    print('Meta não batida!')
print('Fim do programa')
```

```python
investimento = float(input('Informe o valor do investimento em Reais:'))
recebimento = float(input('Informe o valor recebido em Reais:'))
rendimento = (recebimento - investimento) / investimento
meta1 = 0.05
meta2 = 0.2
taxa = 0


if rendimento < meta1:
    taxa = 0
    print('O rendimento foi menor do que a meta, não haverá cobrança de taxa')
    if rendimento >= meta1:
        taxa = 0.02
        print('O rendimento foi superior a 5%, a taxa cobrada será de 2%')
        if rendimento >= meta2:
            taxa = 0.04
            print('O rendimento foi superior a 20%, a taxa cobrada será de 4%')

print('Fim do programa!')
```
